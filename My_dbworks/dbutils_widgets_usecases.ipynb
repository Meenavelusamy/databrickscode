6. dbutils widget usecase - Create dropdown and text widgets... <br>
TODO: Create a dropdown widget for environment (dev, qa, prod) using <br>
TODO: Create a text widget for owner **name**



```python
# 1. Create a dropdown widget for environment
dbutils.widgets.dropdown(
    name="environment",
    defaultValue="dev",
    choices=["dev", "qa", "prod"],
    label="Select Environment"
)

# 2. Create a text widget for owner name
dbutils.widgets.text(
    name="owner_name",
    defaultValue="",
    label="Owner Name"
)
```

7. dbutils widget Usecase – Read Widget Values environment and owner and print in the screen

```python
# 1. Read values from the widgets created earlier
env = dbutils.widgets.get("environment")
owner = dbutils.widgets.get("owner_name")

# 2. Print the retrieved values to the screen
print(f"Environment: {env}")
print(f"Owner Name: {owner}")
```

8. dbutils widget Usecase – Move the above processed File to Archive

```python
# 1. Read widget values (optional, useful for tagging archived files)
env = dbutils.widgets.get("environment")
owner = dbutils.widgets.get("owner_name")

# 2. Define source (processed) and destination (archive) paths
source_file = f"{processed_path}/patients.csv"
archive_file = f"{archive_path}/patients_{env}.csv"

# 3. Move the file from processed to archive
dbutils.fs.mv(source_file, archive_file)

print(f"Moved file to {archive_file} (Environment: {env}, Owner: {owner})")

# 4. Verify the file is now in archive and removed from processed
display(dbutils.fs.ls(archive_path))
```
